```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv("../dataset/Cleaned/Cleaned_Master_Environmental_Impact_Dataset.csv")
```

```python
total_countries = df["Country"].nunique()
total_cities = df["City"].nunique()

average_aqi = round(df["AQI Value"].mean(),2)
maximum_aqi = df["AQI Value"].max()
minimum_aqi = df["AQI Value"].min()

print("Total Countries :", total_countries)
print("Total Cities :", total_cities)
print("Average AQI :", average_aqi)
print("Maximum AQI :", maximum_aqi)
print("Minimum AQI :", minimum_aqi)
```

```python
print("Average CO AQI :", round(df["CO AQI Value"].mean(),2))
print("Average Ozone AQI :", round(df["Ozone AQI Value"].mean(),2))
print("Average NO2 AQI :", round(df["NO2 AQI Value"].mean(),2))
print("Average PM2.5 AQI :", round(df["PM2.5 AQI Value"].mean(),2))
```

```python
plt.figure(figsize=(8,5))

df["AQI Category"].value_counts().plot(kind="bar")

plt.title("AQI Category Distribution")
plt.xlabel("AQI Category")
plt.ylabel("Number of Cities")

plt.show()
```

```python
country = df.groupby("Country")["AQI Value"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

country.plot(kind="bar")

plt.title("Top 10 Countries by Average AQI")
plt.xlabel("Country")
plt.ylabel("Average AQI")

plt.xticks(rotation=45)

plt.show()
```

```python
top10 = df.sort_values("AQI Value",ascending=False).head(10)

plt.figure(figsize=(10,6))

plt.barh(top10["City"],top10["AQI Value"])

plt.gca().invert_yaxis()

plt.title("Top 10 Most Polluted Cities")
plt.xlabel("AQI Value")

plt.show()
```

```python
pollutants = [
    "CO AQI Value",
    "Ozone AQI Value",
    "NO2 AQI Value",
    "PM2.5 AQI Value"
]

plt.figure(figsize=(8,5))

df[pollutants].mean().plot(kind="bar")

plt.title("Average Pollutant Levels")
plt.ylabel("Average AQI")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["AQI Value"],bins=20)

plt.title("AQI Distribution")
plt.xlabel("AQI Value")
plt.ylabel("Frequency")

plt.show()
```
