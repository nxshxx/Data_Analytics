```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside notebook
%matplotlib inline
```

```python
df = pd.read_csv("../dataset/Cleaned/Cleaned_Master_Environmental_Impact_Dataset.csv")
```

```python
print("Rows, Columns:", df.shape)
```

```python
print("Total Countries:", df["Country"].nunique())
print("Total Cities:", df["City"].nunique())
```

```python
print("Average AQI:", round(df["AQI Value"].mean(), 2))
print("Maximum AQI:", df["AQI Value"].max())
print("Minimum AQI:", df["AQI Value"].min())

print("Average CO AQI:", round(df["CO AQI Value"].mean(), 2))
print("Average Ozone AQI:", round(df["Ozone AQI Value"].mean(), 2))
print("Average NO2 AQI:", round(df["NO2 AQI Value"].mean(), 2))
print("Average PM2.5 AQI:", round(df["PM2.5 AQI Value"].mean(), 2))
```

```python
df["AQI Category"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["AQI Category"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("AQI Category Distribution")
plt.xlabel("AQI Category")
plt.ylabel("Number of Cities")
plt.xticks(rotation=45)
plt.show()
```

```python
top10 = df.sort_values("AQI Value", ascending=False).head(10)

top10
```

```python
least10 = df.sort_values("AQI Value", ascending=True).head(10)

least10
```

```python
plt.figure(figsize=(10,6))

plt.barh(least10["City"], least10["AQI Value"])

plt.xlabel("AQI Value")
plt.ylabel("City")
plt.title("Top 10 Least Polluted Cities")

plt.gca().invert_yaxis()

plt.show()
```

```python
country_aqi = (
    df.groupby("Country")["AQI Value"]
      .mean()
      .sort_values(ascending=False)
)

country_aqi.head(10)
```

```python
plt.figure(figsize=(12,6))

country_aqi.head(10).plot(kind="bar")

plt.title("Top 10 Countries with Highest Average AQI")
plt.xlabel("Country")
plt.ylabel("Average AQI")
plt.xticks(rotation=45)

plt.show()
```

```python
pollutants = [
    "CO AQI Value",
    "Ozone AQI Value",
    "NO2 AQI Value",
    "PM2.5 AQI Value"
]

df[pollutants].mean()
```

```python
plt.figure(figsize=(8,5))

df[pollutants].mean().plot(kind="bar")

plt.title("Average Pollutant Levels")
plt.ylabel("Average AQI Value")
plt.xticks(rotation=20)

plt.show()
```

```python
corr = df[
    [
        "AQI Value",
        "CO AQI Value",
        "Ozone AQI Value",
        "NO2 AQI Value",
        "PM2.5 AQI Value"
    ]
].corr()

corr
```

```python
import seaborn as sns

plt.figure(figsize=(8,6))

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Between AQI and Pollutants")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["AQI Value"], bins=20)

plt.title("Distribution of AQI Values")
plt.xlabel("AQI Value")
plt.ylabel("Frequency")

plt.show()
```

The cities with the highest and lowest AQI were identified.
Country-level AQI performance was analyzed.
Pollutant averages were compared.
Correlation analysis showed which pollutants most strongly influence AQI.
The analysis provides insights that can support environmental monitoring and pollution-control strategies.
